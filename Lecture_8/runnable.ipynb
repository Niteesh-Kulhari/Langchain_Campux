```python
import random
class LLM:

    def __init__(self):
        print('LLM created')
    
    def predict(self, prompt):

        response_list = [
            'Delhi is the capital of India',
            'IPL is a cricket league',
            'AI stands for artificial intelligence'
        ]

        return{'response': random.choice(response_list)}
```

```python
llm = LLM()
```

```python
llm.predict('Capital')
```

```python
class PromptTemplate:

    def __init__(self, template, input_variables):
        self.template = template
        self.input_variables = input_variables

    def format(self, input_dict):
        return self.template.format(**input_dict)
```

```python
template = PromptTemplate(
    template='Write a poem about {topic}',
    input_variables=['topic']
)

input_dict = {
    'topic': 'AI'
}

prompt = template.format(input_dict)

# print(res)

```

```python
llm = LLM()
llm.predict(prompt)
```

```python
class LLMCHAINN:

    def __init__(self, llm, prompt):
        self.llm = llm
        self.prompt = prompt

    def run(self, input_dict):
        final_prompt = prompt.format(input_dict)
        result = self.llm.predict(final_prompt)

        return result['response']
```

```python
template = PromptTemplate(
    template='Write a {length} summary about {topic}',
    input_variables=['length', 'topic']
)
```

```python
llm = LLM()
```

```python
chain = LLMCHAINN(llm, template)
chain.run(
    {'length': 'short', 'topic': 'NLP'}
)
```

# Turning the classes in Runnables

```python
from abc import ABC, abstractmethod

class Runnable(ABC):

    @abstractmethod
    def invoke(input_data):
        pass
```

```python
import random
class LLM(Runnable):

    def __init__(self):
        print('LLM created')

    def invoke(self, prompt):
    
            response_list = [
                'Delhi is the capital of India',
                'IPL is a cricket league',
                'AI stands for artificial intelligence'
            ]
    
            return{'response': random.choice(response_list)}
    
    def predict(self, prompt):

        response_list = [
            'Delhi is the capital of India',
            'IPL is a cricket league',
            'AI stands for artificial intelligence'
        ]

        return{'response': random.choice(response_list)}
```

```python
llm = LLM()
```

```python
class PromptTemplate(Runnable):

    def __init__(self, template, input_variables):
        self.template = template
        self.input_variables = input_variables

    def invoke(self, input_dict):
        return self.template.format(**input_dict)

    def format(self, input_dict):
        return self.template.format(**input_dict)
```

```python
class RunnableConnector(Runnable):

    def __init__(self, runnable_list):
        self.runnable_list = runnable_list

    def invoke(self, input_data):

        for runnable in self.runnable_list:
            input_data = runnable.invoke(input_data)
            print(input_data)

        return input_data
```

```python
class STRparser(Runnable):

    def __init__(self):
        pass

    def invoke(self, input_data):
        return input_data['response']
```

```python
template = PromptTemplate(
    template='Write a {length} poem about {topic}',
    input_variables=['length', 'topic']
)

llm = LLM()
parser = STRparser()

chain = RunnableConnector([template, llm, parser])

chain.invoke({'length': 'short', 'topic': 'AI'})
```

```python
template1 = PromptTemplate(
    template='Write a joke about {topic}',
    input_variables=['topic']
)

template2 = PromptTemplate(
    template='Write a summary about the following joke {response}',
    input_variables=['resposne']
)

llm = LLM()
parser = STRparser()
```

```python
chain1 = RunnableConnector([template1, llm])
chain2 = RunnableConnector([template2, llm, parser])
```

```python
final_chain = RunnableConnector([chain1, chain2])
final_chain.invoke({'topic': 'AI'})
```
